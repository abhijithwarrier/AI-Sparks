**PYTHON SCRIPT TO _LOAD A TEXT DOCUMENT AND USE A PRE-TRAINED QA MODEL TO ANSWER QUESTIONS._** 🐍📄️

This script shows how to apply a **question-answering pipeline** from Hugging Face. We’ll provide a passage of text as *context* and ask natural language questions. The model extracts the most relevant answer directly from the text.

### 📦 Install & Import Libraries
We’ll use Hugging Face’s `transformers` library for the QA pipeline.

```python
!pip install transformers

from transformers import pipeline
```

### 📖 Load a Sample Text Document
In practice, you could load text from a `.txt` or PDF.
Here we’ll define a short passage directly in Python.

```python
context = """
Python is a powerful programming language widely used for AI and machine learning.
It is popular because of its simple syntax, strong community support, and rich ecosystem of libraries like TensorFlow, PyTorch, and scikit-learn.
Python’s versatility makes it suitable for web development, data science, automation, and more.
"""
```

### ❓ Define the QA Pipeline
We’ll use a pre-trained model fine-tuned on SQuAD (`distilbert-base-cased-distilled-squad`).

```python
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
```

### 💬 Ask Questions from the Text
The model extracts answers directly from the given context.

```python
questions = [
    "Why is Python popular?",
    "Which libraries are mentioned for machine learning?",
    "What are some applications of Python?"
]

for q in questions:
    result = qa_pipeline(question=q, context=context)
    print(f"Q: {q}")
    print(f"A: {result['answer']}\n")
```

### 📝 Notes
- For larger documents, you may need to split text into chunks before passing to the model.
- Works best on **fact-based questions** where answers are explicitly in the context.
- For more advanced use cases, explore **retrieval-augmented QA** with external document search.
